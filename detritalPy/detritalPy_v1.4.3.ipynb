## Plot and export maximum depositional age (MDA) calculations
Calculate the maximum depositional age (MDA) for a sample or group of samples. Results can be exported as individual PDF figures for each sample or sample group. Results may also be exported to a CSV file.  
<b><i>Note: </i></b>A new MDA plotting function was introduced in notebook version v1.4.3 that incorporates a wider range MDA calculations and plotting parameters.  
<b><i>Warning: </i></b>MDA calculations that preferentially select the youngest grains (the youngest single grain, YSG, or youngest clusters) are prone to underestimating the true MDA. I recommend referring to Oleson et al. (in review): Journal of Sedimentary Research for a comparison of the accuracy of different MDA calculation methods.

| MDA method | Abbreviation | N* | Method | Selected reference |
| :--- | :--- | :--- | :--- | :---- |
| Youngest single grain | YSG | 1 | Single analysis | Dickinson and Gehrels (2009) |
| Youngest three grains | Y3Za | 3 | Weighted mean | Zhang et al. (2016) |
| Youngest three grains (2σ overlap) | Y3Zo | 3 | Weighted mean | Ross et al. (2017) |
| Youngest cluster (1σ overlap) | YC1s | ≥2 | Weighted mean | Dickinson and Gehrels (2009) |
| Youngest cluster (2σ overlap) | YC2s | ≥3 | Weighted mean | Dickinson and Gehrels (2009) |
| Youngest probability peak | YPP | ≥2 | Peak age | Dickinson and Gehrels (2009) |
| Youngest statistical population | YSP | ≥2 | Weighted mean | Coutts et al. (2019) |
| 𝛕 method | tauMethod | ≥3 | Weighted mean | Barbeau et al. (2009) |
| Youngest Gaussian fit | YGF | ≥3 | Statistical fit | Saylor et al. (2023) |

*Minimum number of analyses can be specified  
Table modified from Sharman and Malkowski (2020): <i>Earth-Science Reviews</i>  
YGF is experimental - use with caution.
See MDAfuncs.py for documentation for each MDA method

## Multi-dimensional scaling
Create a multi-dimensional scaling (MDS) plot for individual samples or groups of samples. <i><b>Revised and updated in detritalpy v1.4.15, scikit-learn v1.7+ required.</b></i>

```python
import sklearn
print(sklearn.__version__) # scikit-learn v1.7+ required
```

```python
sampleList = ['11-Escanilla','12-Escanilla','10-Sobrarbe','7-Guaso','13-Guaso','5-Morillo','6-Morillo','14AB-M02','14AB-A04','14AB-A05','4-Ainsa','14AB-A06','15AB-352','15AB-118','15AB-150','3-Gerbe','14AB-G07','2-Arro','1-Fosado','14AB-F01']

ages, errors, numGrains, labels = dFunc.sampleToData(sampleList, main_byid_df, sigma = '1sigma');
```

Run the model. Note that this may take a while, depending on the number of initializations (`n_init`) chosen. Increasing the number of initializations increases the likelihood of finding a more optimal fit (i.e., lower stress), particularly for non-metric MDS. For non-metric MDS, set `init='random'` to use `n_init` random non-metric starts, or set `init='metric'` to use the best of `n_init` metric MDS starts to initialize a single non-metric fit. The `init` argument is ignored when `metric=True`.

```python
model = dFunc.MDS_class(ages, errors, labels, sampleList, metric=False, init='metric', criteria='Vmax', bw=3, n_init=100, 
                        max_iter=1000, x1=0, x2=4500, xdif=1, min_dim=1, max_dim=3, dim=2, normalized_stress=True)
```

<b><i>Note:</i></b> If normalized_stress='auto', then the Stress-1 (Kruskal, 1964) is returned for non-metric MDS and raw stress is returned for metric MDS. Set normalized_stress=True to return Stress-1 for metric MDS as well (requires v1.7+). Please refer to the sklearn.manifold.MDS documentation for additional details.

After the model has been run, a number of figures can be generated, in any order.

The QQ matrix plots each sample CDF or cumulative PDPs/KDEs against the others. A perfect match falls along the dashed line. <i>Note: This is not recommended for large datasets (e.g., >20 samples or sample groups)</i>

```python
model.QQplot(figsize=(12,12), savePlot=False, fileName='QQplot.pdf', halfMatrix=True)
```

A heatmap of the sample dissimilarity matrix gives an indication of the data that is going into the MDS algorithm

```python
model.heatMap(figsize=(10,10), savePlot=False, fileName='HeatMapPlot.pdf', plotValues=True,
              plotType='dissimilarity', fontsize=10)
```

By setting the keyword argument plotType to equal 'distance', we can plot the Euclidean distance between sample pairs on the MDS plot. There should be a general correlation with the heat map above, as samples that are more different should be farther apart.

```python
model.heatMap(figsize=(10,10), savePlot=False, fileName='HeatMapPlot.pdf', plotValues=True, plotType='distance', fontsize=10)
```

A stress plot gives an indication of the goodness-of-fit and how this varies depending on how many dimensions are modeled

```python
model.stressPlot(figsize=(6,6), savePlot=False, fileName='stressPlot.pdf')
```

A Shepard plot compares Euclidean distance on the MDS plot against the dissimilarity metric. Ideally, sample pairs that are far apart on the MDS plot (large distance) will also be the most dissimilar, and vice versa. The amount of scatter gives a sense of the stress value - a lot of scatter means that there's a lot of variance in how well distance on the MDS plot characterizes sample dissimilarity.

```python
model.shepardPlot(figsize=(6,6), savePlot=False, fileName='shepardPlot.pdf', plotOneToOneLine=False)
```

The MDS plot is a depiction of sample similarity and dissimilarity (refer to Vermeesch, 2013: Chemical Geology for a more complete description).

```python
model.MDSplot(figsize=(6,6), savePlot=False, fileName='MDSplot.pdf', plotLabels=True, equalAspect=False, 
              stressType='sklearn')
```

Note that individual samples can be colored according to a categorical variable (column name in the 'Samples' table)

```python
model.MDSplot(figsize=(6,6), savePlot=False, fileName='MDSplot.pdf', plotLabels=True, colorBy='Unit', 
              df=main_byid_df, equalAspect=False, stressType='sklearn')
```
